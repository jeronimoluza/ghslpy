```python
import matplotlib.pyplot as plt
import geopandas as gpd
import mapclassify
import ghslpy
from shapely import wkt
```

```python
# Define study area (Lagos, Nigeria)
buf = 60_000
p = "POINT(3.3642891130747983 6.482301819593587)" # Lagos, Nigeria
# p = "POINT(8.51259507981301 12.002657077147319)" # Kano, Nigeria
# p = "POINT(5.623935754478583 6.338754647182569)" # Benin City, Nigeria
point = gpd.GeoDataFrame([wkt.loads(p)], columns=["geometry"], crs=4326)
buffered = point.copy()
buffered["geometry"] = buffered.geometry.to_crs(3857).buffer(buf).to_crs(4326)
```

```python
# Download GHSL data

# epochs = [1975, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025]
# epochs = [1975, 2000, 2020, 2025]
epochs = [1975, 2025]

data = ghslpy.download(
    products=["GHS-POP", "GHS-BUILT-S"],
    epoch=epochs,
    resolution="1000m",
    region=buffered,
)

print("Data downloaded successfully!")
```

```python
variable_labels = {
    'GHS_POP': 'Population',
    'GHS_BUILT': 'Built-Up Surface Area (m2)'
}
```

```python
extent = wkt.loads('POLYGON((2.8388 7.1335, 4.0805 7.1335, 4.0805 6.3569, 2.8388 6.3569, 2.8388 7.1335))')
```

```python
built25 = data.GHS_BUILT.sel(time='2025-01-01')
pop25 = data.GHS_POP.sel(time='2025-01-01')
```

```python
built_bins = [
    1000, 10000, 50000, 100000, 500000, 800000
]

pop_bins = [
    100, 1000, 5000, 10000, 50000, 110000
]
```

```python
built_scheme={"bins":built_bins}
pop_scheme={"bins":pop_bins}
```

```python
dates = [f"{epoch}-01-01" for epoch in epochs]
```

```python
vector_b25 = ghslpy.vectorize(data.GHS_BUILT.sel(time='2025-01-01').to_dataset(name='built2025'))
vector_p25 = ghslpy.vectorize(data.GHS_POP.sel(time='2025-01-01').to_dataset(name='pop2025'))
```

```python
import folium.plugins
```

```python
m1 = vector_b25.dropna(subset='built2025').explore(
    'built2025',
    cmap='Reds',
    scheme='UserDefined',
    prefer_canvas=True,
    tiles="CartoDB Positron",
    style_kwds = {"stroke":False, "fillOpacity": .7},
    classification_kwds = built_scheme,
    legend_kwds = {
        "caption": 'Built-Up Surface Area (m2) in 2025',
        "colorbar": False,
        "fmt": "{:,.0f}"
    }

)

folium.plugins.Fullscreen(
    position="topright",
    title="Expand me",
    title_cancel="Exit me",
    force_separate_button=True,
).add_to(m1)
folium.LayerControl().add_to(m1)
m1.save('map.html')
m1
```

```python
m = folium.Map(location=(6.5, 3.5), zoom_start=9)

layer_left = vector_b25.dropna(subset='built2025').explore(
    'built2025',
    cmap='Reds',
    scheme='UserDefined',
    #prefer_canvas=True,
    tiles="CartoDB Positron",
    style_kwds = {"stroke":False, "fillOpacity": .7},
    classification_kwds = built_scheme,
    legend_kwds = {
        "caption": 'Built-Up Surface Area (m2) in 2025',
        "colorbar": False,
        "fmt": "{:,.0f}"
    }
)

layer_right = vector_p25.dropna(subset='pop2025').explore(
    'pop2025',
    cmap='Blues',
    scheme='UserDefined',
    #prefer_canvas=True,
    tiles="CartoDB Positron",
    style_kwds = {"stroke":False, "fillOpacity": .7},
    classification_kwds = pop_scheme,
    legend_kwds = {
        "caption": 'Population in 2025',
        "colorbar": False,
        "fmt": "{:,.0f}"
    },
)

sbs = folium.plugins.SideBySideLayers(layer_left=layer_left, layer_right=layer_right)

layer_left.add_to(m)
layer_right.add_to(m)
sbs.add_to(m)

m
```
